```python
def obter_opcao(prompt, opcoes_validas):
    while True:
        try:
            valor = int(input(prompt))
            if valor in opcoes_validas:
                return valor
            print("Opção inválida! Tente novamente.")
        except ValueError:
            print("Entrada inválida! Digite apenas números inteiros.")

def cadastrar_entrevistado(numero):
    print(f"\n--- Entrevistado #{numero} ---")

    print("\nEm quem você pretende votar?")
    print("1 - Candidato A\n2 - Candidato B\n3 - Candidato C\n9 - Indeciso\n0 - Branco/Nulo")
    voto = obter_opcao("Opção: ", [1, 2, 3, 9, 0])

    print("\nQual a sua renda familiar?")
    print("1 - Até 2 salários mínimos\n2 - De 2 a 5 salários mínimos\n3 - Mais de 5 salários mínimos")
    renda = obter_opcao("Opção: ", [1, 2, 3])

    print("\nQual o seu sexo?")
    print("1 - Masculino\n2 - Feminino")
    sexo = obter_opcao("Opção: ", [1, 2])

    print("\nQual a sua faixa etária?")
    print("1 - 16 a 34 anos\n2 - 35 a 59 anos\n3 - 60 anos ou mais")
    idade = obter_opcao("Opção: ", [1, 2, 3])

    print("\nQual o seu grau de escolaridade?")
    print("1 - Ensino Fundamental\n2 - Ensino Médio\n3 - Ensino Superior")
    escolaridade = obter_opcao("Opção: ", [1, 2, 3])

    print("\nQual a sua cor/raça?")
    print("1 - Branca\n2 - Parda\n3 - Preta")
    raca = obter_opcao("Opção: ", [1, 2, 3])

    print("\nQual a sua religião?")
    print("1 - Católica\n2 - Evangélica\n3 - Outras / Sem religião")
    religiao = obter_opcao("Opção: ", [1, 2, 3])

    return {
        "voto": voto,
        "renda": renda,
        "sexo": sexo,
        "idade": idade,
        "escolaridade": escolaridade,
        "raca": raca,
        "religiao": religiao
    }

def calcular_porcentagem(valor, total):
    if total == 0:
        return 0.0
    return (valor / total) * 100

def exibir_resultados(respostas):
    total = len(respostas)
    if total == 0:
        print("Nenhum dado foi registrado.")
        return

    votos_a = sum(1 for r in respostas if r["voto"] == 1)
    votos_b = sum(1 for r in respostas if r["voto"] == 2)
    votos_c = sum(1 for r in respostas if r["voto"] == 3)
    indecisos = sum(1 for r in respostas if r["voto"] == 9)
    brancos_nulos = sum(1 for r in respostas if r["voto"] == 0)

    print("\n" + "="*35)
    print("===== RESULTADO DA PESQUISA =====")
    print("="*35)
    print(f"Candidato A: {votos_a} votos - {calcular_porcentagem(votos_a, total):.2f}%")
    print(f"Candidato B: {votos_b} votos - {calcular_porcentagem(votos_b, total):.2f}%")
    print(f"Candidato C: {votos_c} votos - {calcular_porcentagem(votos_c, total):.2f}%")
    print(f"Indecisos: {indecisos} votos - {calcular_porcentagem(indecisos, total):.2f}%")
    print(f"Brancos/Nulos: {brancos_nulos} votos - {calcular_porcentagem(brancos_nulos, total):.2f}%")

    print("\n===== RENDA =====")
    print(f"Até 2 salários mínimos: {calcular_porcentagem(sum(1 for r in respostas if r['renda'] == 1), total):.2f}%")
    print(f"2 a 5 salários mínimos: {calcular_porcentagem(sum(1 for r in respostas if r['renda'] == 2), total):.2f}%")
    print(f"Mais de 5 salários mínimos: {calcular_porcentagem(sum(1 for r in respostas if r['renda'] == 3), total):.2f}%")

    print("\n===== SEXO =====")
    print(f"Masculino: {calcular_porcentagem(sum(1 for r in respostas if r['sexo'] == 1), total):.2f}%")
    print(f"Feminino: {calcular_porcentagem(sum(1 for r in respostas if r['sexo'] == 2), total):.2f}%")

    print("\n===== IDADE =====")
    print(f"16 a 34 anos: {calcular_porcentagem(sum(1 for r in respostas if r['idade'] == 1), total):.2f}%")
    print(f"35 a 59 anos: {calcular_porcentagem(sum(1 for r in respostas if r['idade'] == 2), total):.2f}%")
    print(f"60 ou mais: {calcular_porcentagem(sum(1 for r in respostas if r['idade'] == 3), total):.2f}%")

    print("\n===== ESCOLARIDADE =====")
    print(f"Ensino Fundamental: {calcular_porcentagem(sum(1 for r in respostas if r['escolaridade'] == 1), total):.2f}%")
    print(f"Ensino Médio: {calcular_porcentagem(sum(1 for r in respostas if r['escolaridade'] == 2), total):.2f}%")
    print(f"Ensino Superior: {calcular_porcentagem(sum(1 for r in respostas if r['escolaridade'] == 3), total):.2f}%")

    print("\n===== COR/RAÇA =====")
    print(f"Brancos: {calcular_porcentagem(sum(1 for r in respostas if r['raca'] == 1), total):.2f}%")
    print(f"Pardos: {calcular_porcentagem(sum(1 for r in respostas if r['raca'] == 2), total):.2f}%")
    print(f"Pretos: {calcular_porcentagem(sum(1 for r in respostas if r['raca'] == 3), total):.2f}%")

    print("\n===== RELIGIÃO =====")
    print(f"Católicos: {calcular_porcentagem(sum(1 for r in respostas if r['religiao'] == 1), total):.2f}%")
    print(f"Evangélicos: {calcular_porcentagem(sum(1 for r in respostas if r['religiao'] == 2), total):.2f}%")

    votos_validos = votos_a + votos_b + votos_c

    print("\n" + "="*35)
    print("===== CLASSIFICAÇÃO DA ELEIÇÃO =====")
    print("="*35)

    if votos_validos == 0:
        print("Não houveram votos válidos para determinar o resultado.")
    else:
        pct_a = calcular_porcentagem(votos_a, votos_validos)
        pct_b = calcular_porcentagem(votos_b, votos_validos)
        pct_c = calcular_porcentagem(votos_c, votos_validos)

        if pct_a > 50:
            print("O Candidato A possui mais de 50% dos votos válidos.")
            print("Resultado: vitória no primeiro turno.")
        elif pct_b > 50:
            print("O Candidato B possui mais de 50% dos votos válidos.")
            print("Resultado: vitória no primeiro turno.")
        elif pct_c > 50:
            print("O Candidato C possui mais de 50% dos votos válidos.")
            print("Resultado: vitória no primeiro turno.")
        else:
            print("Nenhum candidato possui mais de 50% dos votos válidos.")
            print("Poderá haver segundo turno.")

def main():
    respostas = []
    print("=== INÍCIO DA PESQUISA DE INTENÇÃO DE VOTOS ===")

    while True:
        dados = cadastrar_entrevistado(len(respostas) + 1)
        respostas.append(dados)

        continuar = input("\nDeseja registrar outro entrevistado? (S/N): ").strip().upper()
        if continuar != 'S':
            break

    exibir_resultados(respostas)

if __name__ == "__main__":
    main()
```
