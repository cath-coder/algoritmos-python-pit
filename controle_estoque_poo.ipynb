```python
class Produto:

    # Método construtor
    def __init__(self):
        self.tipo = "produto"

    # Define o identificador
    def setIdentificador(self, identificador):
        self.Identificador = identificador

    # Retorna o identificador
    def getIdentificador(self):
        return self.Identificador

    # Define o preço
    def setPreco(self, preco):
        self.preco = preco

    # Retorna o preço
    def getPreco(self):
        return self.preco

    # Define a quantidade
    def setQuantidade(self, quantidade):
        self.quantidade = quantidade

    # Retorna a quantidade
    def getQuantidade(self):
        return self.quantidade

    # Calcula o valor total do produto em estoque
    def getValorTotalEstoque(self):
        total = self.preco * self.quantidade
        return total



# ==========================
# Programa Principal
# ==========================

# Criando o primeiro veículo
produto1 = Produto()
produto1.setIdentificador('Leite')
produto1.setPreco(7.99)
produto1.setQuantidade(10)

# Criando o segundo veículo
produto2 = Produto()
produto2.setIdentificador('Maçã')
produto2.setPreco(0.99)
produto2.setQuantidade(15)

# Criando o terceiro veículo
produto3 = Produto()
produto3.setIdentificador('Água')
produto3.setPreco(1.99)
produto3.setQuantidade(20)

# ==========================
# Exibindo os resultados
# ==========================

print("===== PRODUTO 1 =====")
print("Produto:", produto1.getIdentificador())
print("Preço:", produto1.getPreco())
print("Quantidade no estoque:", produto1.getQuantidade())
print("Valor total em estoque: R$", produto1.getValorTotalEstoque())
print()


print("===== PRODUTO 2 =====")
print("Produto:", produto2.getIdentificador())
print("Preço:", produto2.getPreco())
print("Quantidade no estoque:", produto2.getQuantidade())
print("Valor total em estoque: R$", produto2.getValorTotalEstoque())
print()

print("===== PRODUTO 3 =====")
print("Produto:", produto3.getIdentificador())
print("Preço: R$", produto3.getPreco())
print("Quantidade no estoque:", produto3.getQuantidade())
print("Valor total em estoque: R$", produto3.getValorTotalEstoque())
print()
```
