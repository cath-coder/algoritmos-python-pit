```python
class Livro:
    def __init__(self, titulo: str, autor: str, ano_publicacao: int, isbn: str):
        self.titulo = titulo
        self.autor = autor
        self.ano_publicacao = ano_publicacao
        self.isbn = isbn
        self.disponivel = True

    def emprestar(self):
        if self.disponivel:
            self.disponivel = False
            print(f"O livro '{self.titulo}' foi emprestado com sucesso.")
        else:
            print(f"Não é possível emprestar: O livro '{self.titulo}' já está emprestado.")

    def devolver(self):
        if not self.disponivel:
            self.disponivel = True
            print(f"O livro '{self.titulo}' foi devolvido e agora está disponível.")
        else:
            print(f"O livro '{self.titulo}' já se encontra na biblioteca.")

    def exibir_informacoes(self):
        status = "Disponível" if self.disponivel else "Emprestado"
        print(f"--- Informações do Livro ---")
        print(f"Título: {self.titulo}")
        print(f"Autor: {self.autor}")
        print(f"Ano: {self.ano_publicacao}")
        print(f"ISBN: {self.isbn}")
        print(f"Situação: {status}\n")

# Criando os objetos
livro1 = Livro("Engenharia de Software", "Ian Sommerville", 2019, "978-8543024974")
livro2 = Livro("Clean Code", "Robert C. Martin", 2008, "978-0132350884")
livro3 = Livro("O Algoritmo Mestre", "Pedro Domingos", 2017, "978-8535929652")

# Teste
livro1.exibir_informacoes()
```
