```python
import random

energia = 100
total_itens = 5
itens_coletados = 0

print(f"Voce esta na Caverna da PIT e precisa coletar {total_itens} itens para escapar.")
print(f"Explorar a caverna consome energia a cada passo. Energia inicial: {energia}\n")

while energia > 0 and itens_coletados < total_itens:
    try:
        passos = int(input("Digite a quantidade de passos que deseja andar (1-20): "))
    except ValueError:
        print("Entrada invalida. Digite apenas numeros inteiros.\n")
        continue

    if passos < 1:
        print("Voce nao pode ficar sem andar ou voltar.\n")
        continue
    if passos > 20:
        print("Voce nao enxerga longe o suficiente para caminhar tanto.\n")
        continue

    energia -= passos
    print(f"Voce gastou {passos} de energia e agora tem {energia} restante.")

    if random.randint(1, 100) <= 30:
        itens_coletados += 1
        print(f"Voce encontrou um item. Itens coletados: {itens_coletados} de {total_itens}\n")
    else:
        print("Voce nao encontrou nenhum item desta vez.\n")

if itens_coletados == total_itens:
    print("Parabens! Voce encontrou todos os itens e conseguiu sair da caverna!")
elif energia <= 0:
    print("Sua energia acabou. Voce morreu de fome na caverna.")
```
