# Lista de Exercícios 2 — Algoritmos de Ordenação

## Parte 1 — Questões Teóricas

### 1. Diferença entre algoritmo estável e não estável
* **Estável:** Preserva a ordem relativa de elementos com valores iguais. Se $A$ e $B$ possuem o mesmo valor e $A$ aparece antes de $B$ na entrada, $A$ continuará vindo antes de $B$ no array ordenado. Exemplo: *Insertion Sort*.
* **Não Estável:** Não garante a manutenção da ordem relativa dos elementos de valor igual após a ordenação. Exemplo: *Selection Sort*.

---

### 2. Tabela de Complexidade

| Algoritmo | Melhor Caso | Pior Caso | Estável? |
| :--- | :---: | :---: | :---: |
| **Bubble Sort** | $O(n)$ | $O(n^2)$ | Sim |
| **Selection Sort** | $O(n^2)$ | $O(n^2)$ | Não |
| **Insertion Sort** | $O(n)$ | $O(n^2)$ | Sim |
| **Merge Sort** | $O(n \log n)$ | $O(n \log n)$ | Sim |
| **Quicksort** | $O(n \log n)$ | $O(n^2)$ | Não |

---

### 3. Selection Sort vs. Insertion Sort no melhor caso
* **Selection Sort:** É sempre $O(n^2)$ pois percorre todo o restante da lista em busca do menor elemento a cada iteração, realizando todas as comparações possíveis independentemente do array estar ordenado.
* **Insertion Sort:** No melhor caso (array já ordenado), cada elemento é comparado apenas uma vez com seu predecessor e a inserção ocorre imediatamente sem deslocamentos, atingindo $O(n)$.

---

### 4. Dividir para Conquistar (Merge Sort vs. Quicksort)
A estratégia consiste em dividir o problema em subproblemas menores, resolvê-los de forma recursiva e combinar os resultados.
* **Merge Sort:** O trabalho principal ocorre na **combinação** (intercalação das sublistas ordenadas).
* **Quicksort:** O trabalho principal ocorre na **divisão** (particionamento ao redor do pivô).

---

### 5. O Pivô no Quicksort
* **Pivô:** Elemento de referência usado para dividir a lista entre elementos menores/iguais e maiores que ele.
* **Pior Caso:** Ocorre quando o pivô escolhido é sempre o maior ou menor elemento da sublista, dividindo a lista de forma totalmente desequilibrada ($1$ e $n-1$ elementos), o que eleva a profundidade de recursão para $O(n)$ e a complexidade para $O(n^2)$.

```python
# Parte 2 — Implementação dos Algoritmos


def bubble_sort(lista):
  n = len(lista)
  for i in range(n):
    trocou = False
    for j in range(0, n - i - 1):
      if lista[j] > lista[j + 1]:
        lista[j], lista[j + 1] = lista[j + 1], lista[j]
        trocou = True
    if not trocou:
      break
  return lista


def selection_sort(lista):
  n = len(lista)
  for i in range(n):
    idx_min = i
    for j in range(i + 1, n):
      if lista[j] < lista[idx_min]:
        idx_min = j
    lista[i], lista[idx_min] = lista[idx_min], lista[i]
  return lista


def insertion_sort(lista):
  for i in range(1, len(lista)):
    chave = lista[i]
    j = i - 1
    while j >= 0 and lista[j] > chave:
      lista[j + 1] = lista[j]
      j -= 1
    lista[j + 1] = chave
  return lista


def merge(esquerda, direita):
  resultado = []
  i = j = 0
  while i < len(esquerda) and j < len(direita):
    if esquerda[i] <= direita[j]:
      resultado.append(esquerda[i])
      i += 1
    else:
      resultado.append(direita[j])
      j += 1
  resultado.extend(esquerda[i:])
  resultado.extend(direita[j:])
  return resultado


def merge_sort(lista):
  if len(lista) <= 1:
    return lista
  meio = len(lista) // 2
  esquerda = merge_sort(lista[:meio])
  direita = merge_sort(lista[meio:])
  return merge(esquerda, direita)


# --- Testes ---
arr1 = [64, 34, 25, 12, 22, 11, 90]
arr2 = [5, 2, 8, 1, 9]

print("Bubble Sort:", bubble_sort(arr1.copy()))
print("Selection Sort:", selection_sort(arr1.copy()))
print("Insertion Sort:", insertion_sort(arr1.copy()))
print("Merge Sort:", merge_sort(arr1.copy()))
```

### 3. Quicksort — Partição de Lomuto (`[8, 3, 5, 1, 9, 2]`, Pivô = 2)

| Elemento comparado | Comparação com o pivô (2) | Ação |
| :---: | :---: | :---: |
| **8** | $8 \le 2$ (Falso) | Nenhum ajuste |
| **3** | $3 \le 2$ (Falso) | Nenhum ajuste |
| **5** | $5 \le 2$ (Falso) | Nenhum ajuste |
| **1** | $1 \le 2$ (Verdadeiro) | Troca `1` com `8` $\rightarrow$ `[1, 3, 5, 8, 9, 2]` |
| **9** | $9 \le 2$ (Falso) | Nenhum ajuste |

* **Troca Final:** O pivô `2` é trocado com o `3`.
* **Array Resultante:** `[1, 2, 5, 8, 9, 3]`
* **Índice Final do Pivô:** `1`

```python
# Parte 4 — Desafio (Quicksort com Lomuto)


def particiona(lista, baixo, alto):
  pivo = lista[alto]
  i = baixo - 1
  for j in range(baixo, alto):
    if lista[j] <= pivo:
      i += 1
      lista[i], lista[j] = lista[j], lista[i]
  lista[i + 1], lista[alto] = lista[alto], lista[i + 1]
  return i + 1


def quicksort(lista, baixo=0, alto=None):
  if alto is None:
    alto = len(lista) - 1
  if baixo < alto:
    pi = particiona(lista, baixo, alto)
    quicksort(lista, baixo, pi - 1)
    quicksort(lista, pi + 1, alto)
  return lista


# Teste
arr_desafio = [8, 3, 5, 1, 9, 2]
print("Quicksort ordenado:", quicksort(arr_desafio.copy()))
```

### Pesquisa sobre o Timsort
* **Por que combina Insertion Sort e Merge Sort?** O Timsort divide o array em pequenos blocos (*runs*, tipicamente de tamanho 32 a 64). Ele ordena cada bloco individualmente com **Insertion Sort** e depois combina as partes com a lógica do **Merge Sort**.
* **Por que o Insertion Sort é útil nele?** Para sublistas pequenas, o Insertion Sort possui pouquíssimo overhead e é extremamente rápido, evitando chamadas recursivas desnecessárias do Merge Sort em trechos pequenos.
