```python
salario = float(input('Digite se salário: R$ '))

print('Essenciais (50%): R$', salario * 0.50)
print('Lazer (20%): R$', salario * 0.20)
print('Investimento (30%): R$', salario * 0.30)

print('\n--- SIMULADOR DE INVESTIMENTO ---')

aporte_inicial = float(input('Aporte inicial: R$ '))
aporte_mensal = float(input('Aporte mensal: R$ '))
taxa_anual = float(input('Taxa de juros anual: '))
anos = int(input('Quantidade de anos investidos:'))

taxa_mensal = ((1 + taxa_anual) ** (1/12)) - 1
meses = anos * 12

saldo = aporte_inicial
total_investido = aporte_inicial

for m in range(1, meses + 1):
  saldo = (saldo * (1 + taxa_mensal)) + aporte_mensal
  total_investido += aporte_mensal

  if m % 12 == 0:
    print(f'Ano {m // 12}: R$ {saldo:.2f}')

print('\n-----------------------------------')
print('Total do seu bolso:', total_investido)
print('Total acumulado com juros:', round(saldo,2))
print('Rendimento aproximado por mês:', round(saldo * 0.01, 2))
print('\n-----------------------------------')
```
